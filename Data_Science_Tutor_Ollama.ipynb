```python
import ollama
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
```

### Prompt for Tutor

```python
SYSTEM_PROMPT = '''
You are an expert Data Science Tutor which explains things in a simplified manner.

Rules:
- Explain in simple language.
- Give intuition.
- Give a real-world example.
- Provide Python code whenever required.
- Explain mathematics only if needed.
- End every answer with one quiz question.
'''

messages=[{"role":"system","content":SYSTEM_PROMPT}]
```

#### UI

```python
title=widgets.HTML("<h1>📘 Data Science Tutor </h1>")

topic=widgets.Dropdown(options=["Machine Learning","Deep Learning","Python","Statistics","SQL","NLP","Prompt Engineering"],
                       description="Topic:")

difficulty=widgets.Dropdown(options=["Beginner","Intermediate","Advanced"],description="Level:")

question=widgets.Text(description="Question:",layout=widgets.Layout(width="90%"))

ask_btn=widgets.Button(description="Ask AI",button_style="success")

clear_btn=widgets.Button(description="Clear Chat",button_style="danger")

output=widgets.Output(layout={'border':'1px solid gray','padding':'10px'})
```

#### Backend

```python
def ask_ai(_):
    with output:
        clear_output()
        q=question.value.strip()
        if not q:
            print("Please enter a question.")
            return
        prompt=f"Topic: {topic.value}\nDifficulty: {difficulty.value}\nQuestion: {q}"
        
        messages.append({"role":"user","content":prompt})
        
        response=ollama.chat(model="gemma3",messages=messages)
        
        ans=response["message"]["content"]
        
        messages.append({"role":"assistant","content":ans})
        
        display(Markdown(f"## 👨‍🎓 Student\n{q}"))
        display(Markdown("---"))
        display(Markdown(f"## 🤖 Tutor\n{ans}"))
        question.value=""

def clear_chat(_):
    global messages
    messages=[{"role":"system","content":SYSTEM_PROMPT}]
    with output:
        clear_output()
        display(Markdown("### Chat Cleared ✅"))

ask_btn.on_click(ask_ai)
clear_btn.on_click(clear_chat)

display(widgets.VBox([
    title,topic,difficulty,question,
    widgets.HBox([ask_btn,clear_btn]),
    output
]))
```
