```python
from groq import Groq
import os

client = Groq(api_key="")
```

#### Sytem Prompt

```python
messages = [
    {
        "role":"system",
        "content":"""
You are an expert Python coding assistant.

Your responsibilities are:
- Answer programming questions
- Explain code line by line
- Debug Python errors
- Suggest better coding practices.
- Write clean and readable code.
- If unsure, clearly say you are unsure instead of making up the answer.
- If the questions are apart from python, say that "The topic is out of my context."

Always explain before giving code.
"""
    }
]
```

```python
print("=" * 60)
print("Python Code Assistant")
print("Type 'exit' to quit")
print("=" * 60)

while True:
    question = input("Please write your question: ")
    if question.lower() == "exit":
        print("Goodbye!!!")
        break

    messages.append(
        {
            "role":"user",
            "content":question
        }
    )

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=messages,
        temperature=0.3
    )

    answer = response.choices[0].message.content

    print("\n🤖 Assistant:\n")
    print(answer)

    messages.append(
        {
            "role": "assistant",
            "content": answer
        }
    )
```
